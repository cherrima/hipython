```python
import os

# 데이터셋 경로 설정
dataset_path = "./datasets/Mnist_numbers"

for split in ['train', 'valid', 'test'] :
    images = os.listdir(f"{dataset_path}/{split}/images")
    labels = os.listdir(f"{dataset_path}/{split}/labels")
    
    print(f"{split}: 이미지 {len(images)}개, 라벨 {len(labels)}개")

print()

with open(f"{dataset_path}/data.yaml", "r") as f:
    print(f.read())
```

```python
from ultralytics import YOLO

# YOLOv8 모델 로드
model = YOLO("yolov8n.pt")

# 데이터셋 경로 설정
data_path = "./datasets/Mnist_numbers/data.yaml"

# 모델 훈련 
# result = model.train(data=data_path, epochs=50, imgsz=640, batch=15)

result = model.train(data=data_path, epochs=50, imgsz=640, batch=15, 
                     project="runs_numbers", name="mnist_exp1", exist_ok=True)
```

```python
model = YOLO("./runs/detect/runs_numbers/mnist_exp1/weights/best.pt")

filename = './datasets/numbers/000881.jpg'

results = model.predict(filename, conf=0.5)

for result in results :
    for box in result.boxes :
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        conf = box.conf[0].item()
        cls = result.names[int(box.cls[0])]
        print(f"{cls} {conf:.4f}")
```
