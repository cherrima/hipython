```python
cur_year = 2024
cur_month = 3
cur_day = 15

weekday = ['일','월', '화', '수', '목', '금', '토']

total_days = 0

def leap_year(year) :
    leap_year = False
    if year % 4 == 0 :
        if year % 100 == 0 :
            if year % 400 == 0 : 
                leap_year = True
        else :
            leap_year = True

    return  leap_year

print(leap_year(2024))
print(leap_year(1200))
print(leap_year(2025))

# 연도계산
for year in range(1, cur_year) :
    total_days += 366 if leap_year(year) else 365

# Month days
    month_days = { 1:31, 2:28, 3:31, 4:30, 5:31, 6:30, 7:31, 8:31, 9:30, 10:31, 11:30, 12:31}
    month_days_leap =  { 1:31, 2:29, 3:31, 4:30, 5:31, 6:30, 7:31, 8:31, 9:30, 10:31, 11:30, 12:31}

if ( leap_year(cur_year) ) :
    mdays = month_days_leap
else :
    mdays = month_days

for month in range(1, cur_month) :
    total_days += mdays[month]

total_days += cur_day

print(f"Total Days : {total_days :,}, Week Day : { total_days % 7 }, Week : {weekday[total_days % 7]}요일")
```

```python
from operator import itemgetter

students = [{'num':1, 'name':'Kim', 'kor':80, 'math':90, 'eng':100},
            {'num':2, 'name':'Park', 'kor':90, 'math':60, 'eng':80},
            {'num':3, 'name':'Hong', 'kor':70, 'math':70, 'eng':80},
            {'num':4, 'name':'Lee', 'kor':70, 'math':70, 'eng':90},
            {'num':5, 'name':'Yun', 'kor':70, 'math':70, 'eng':80}
           ]

totals = []
for student in students :
    student['total'] = student['kor'] + student['math'] + student['eng']
    student['average'] = round(student['total'] / 3, 2)
#    print(student)

# print("-"*90)

#sorted_students = sorted(students, key=itemgetter('total'), reverse=True)
sorted_students = sorted(students, key = lambda student : student['total'], reverse=True)

# 학생별 등수
for idx, student in enumerate(sorted_students) :
    student['rank'] = idx + 1
    print(student)

# 전체 총점 및 평균
student_totals = 0
student_count = len(students)

subject_avg = {"kor":0, "math":0, "eng":0}

for student in students :
    student_totals += student['total']
    for subject in subject_avg :
        subject_avg[subject] += student[subject]

for subject in subject_avg :
    subject_avg[subject] = round(subject_avg[subject] / student_count, 2)

print(f"전체 총점 : {student_totals:,}, 전체 평균 : {round(student_totals/student_count/3, 2)}")
print(f"과목별 평균 : {subject_avg}")
```

```python
# 양수 음수 계수기
# '''
# 양수음수분류기 다양한 개수의 정수를 입력받아, 각 수가 양수인지 음수인지 구분하여 각각의 개수를 출력하는 코드를 작성하시오.

# - 입력으로부터 정수리스트로 변환하는 것은 한 줄 코드로 작성
# - 입력 받은 값 중 0은 무시
# - quit 를 입력하면 프로그램을 중지합니다.
# '''

(counter_even, counter_odd) = (0,0)
(positive, negative) = ([],[])
numbers = []
while True :
    (counter_even, counter_odd) = (0,0)
    (positive, negative) = ([],[])
    numbers = []

    user_input = input("정수 목록을 공백으로 구분하여 입력하세요. 종료하고 싶으면 'q'를 입력하세요 :")
    if len(user_input) < 1 : 
        print("입력받은 숫자가 없습니다.")
        continue 
        
    if user_input.lower()[0] == 'q' : 
        print("프로그램 종료.")
        break

    user_list = user_input.split() # 기본 공백
    for item in user_list :

        try:
            num = int(item.strip())
        except ValueError:
            continue
            
        # if ( num == 0 ) continue
        if ( num != 0 ) :
            numbers.append(num)
            if ( num %2 == 0 ) : counter_even += 1
            else : counter_odd += 1
            
            if ( num > 0 ) : positive.append(num)
            else : negative.append(num)

#  numbers = [int(item) for item in user_list]
#  numbers = [x for x in numbers if x != 0]        
            
        if len(numbers) == 0 :
            print("입력받은 숫자가 없습니다.")

    print(numbers)
    print(f"Positive numbers : {positive}, Negative Numbers : {negative}")
    print(f"Even number count : {counter_even}, Odd Number count : {counter_odd}")
    print(f"Positive number count : {len(positive)}, Negative Number count : {len(negative)}, Total Sum : {sum(numbers)}")
```
