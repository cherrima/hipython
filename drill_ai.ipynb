```python
from openai import OpenAI
from dotenv import load_dotenv
import os

# print(os.getenv("PATH"))

# .env 파일 로드
load_dotenv()

# 헬퍼 
def query_helper(text, role="user"):
    query = {
        "role": role,
        "content": [{"type":"input_text","text": text}]
    }
 
    return query

# Prompt 입력, 실행, 및 결과 리턴 API 키 읽기
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY")) # OpenAI 서버 연결
print(client)
```
