```python
import seaborn as sns
```

```python
tips_df = sns.load_dataset('tips')
tips_df
```

```python
tips_day_df = tips_df.groupby('day').mean(numeric_only=True)
tips_day_df['tip_per_bill'] = tips_day_df['tip'] / tips_day_df['total_bill']
tips_day_df
```

```python
sns.scatterplot(data=tips_df, x='total_bill', y='tip')
sns.scatterplot(data=tips_df, x='total_bill', y='tip', hue='sex')
```

```python
x = [1,2,3,4,5]
y = [10, 20, 25, 30, 42]

# sns.scatterplot(x=x, y=y)
sns.lineplot(x=x, y=y)
sns.barplot(x=x, y=y)
```

```python
# sns.regplot(x=x, y=y)
# sns.regplot(data=tips_df, x='total_bill', y='tip') 
sns.regplot(x=tips_df['total_bill'], y=tips_df['tip'])
```

```python
sns.pointplot(data=tips_df, x='day', y='tip') 
```

```python
sns.boxplot(x='size', y='tip', data=tips_df)
```

```python
sns.violinplot(x='size', y='tip', data=tips_df)
```

```python
sns.histplot(tips_df['tip'], bins=10)
```

```python
sns.kdeplot(tips_df['tip'], fill=True)
sns.rugplot(tips_df['tip'], color='black')
```

```python
sns.heatmap(tips_df.corr(numeric_only=True), annot= )
```

```python
pivot_df = tips_df.pivot_table('tip', 'day', 'size', aggfunc='count', observed=False)
pivot_df   
```

```python
sns.heatmap(pivot_df, annot=True, cmap='YlGnBu')
```

```python
# Time, Sex or Time, Smoker
pivot_time_sex = tips_df.pivot_table('tip', 'time', 'sex', aggfunc='count', observed=False)
pivot_time_sex
sns.heatmap(pivot_time_sex, annot=True, cmap='YlGnBu', fmt='d') 
```

```python
pivot_time_smoker = tips_df.pivot_table('tip', 'time', 'smoker', aggfunc='count', observed=False)
print(pivot_time_smoker)
print(pivot_time_smoker.index, pivot_time_smoker.index.name)
print(pivot_time_smoker.columns, pivot_time_smoker.columns.name)
sns.heatmap(pivot_time_smoker, annot=True, cmap='YlGnBu', fmt='d') 
```
