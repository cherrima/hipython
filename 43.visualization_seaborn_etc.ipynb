```python
import folium

# 1. 좌표 설정 (위도, 경도)
# 노량진역: 37.5139, 126.9412
# 내방역: 37.4877, 126.9972
noryangjin = [37.5139, 126.9412]
naebang = [37.4877, 126.9972]

# 2. 지도 초기화 (두 지점의 중간쯤으로 설정)
center = [(noryangjin[0] + naebang[0]) / 2, (noryangjin[1] + naebang[1]) / 2]
m = folium.Map(location=center, zoom_start=13)

# 3. 마커 추가 (노량진역)
folium.Marker(
    location=noryangjin,
    popup='노량진역',
    tooltip='노량진역',
    icon=folium.Icon(color='blue', icon='info-sign')
).add_to(m)

# 4. 마커 추가 (내방역)
folium.Marker(
    location=naebang,
    popup='내방역',
    tooltip='내방역',
    icon=folium.Icon(color='red', icon='info-sign')
).add_to(m)

# 5. 직선 경로 표시 (PolyLine)
locations = [noryangjin, naebang]
folium.PolyLine(
    locations=locations,
    color='red',
    weight=5,
    opacity=0.7,
    tooltip='직선거리'
).add_to(m)

# 6. 지도 저장
m.save('./data/subway_line.html')
print("지도가 'subway_line.html' 파일로 저장되었습니다.")

```
