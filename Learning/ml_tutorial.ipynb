```python
import tensorflow as tf
from sklearn import linear_model
import numpy as np
import pandas as pd
```

# Machine Learning

```python
X_train = np.array([1,2,3,4,5,6,7]).reshape(-1,1)
Y_train = np.array([12,17,22,27,32,37,42]).reshape(-1,1)
model = linear_model.LinearRegression() # Model
model.fit(X_train, Y_train) # 훈련
# 예측
model.predict([[9]])
```

# Deep Learning

```python
X_train = np.array([1,2,3,4,5,6,7]).reshape(-1,1)
Y_train = np.array([12,17,22,27,32,37,42]).reshape(-1,1)

inputs = tf.keras.layers.Input(shape=(1,))
outputs = tf.keras.layers.Dense(units=1)(inputs)

model = tf.keras.models.Model(inputs, outputs)
model.compile(loss=tf.keras.losses.MeanSquaredError(), optimizer=tf.keras.optimizers.SGD())
model.fit(X_train, Y_train, epochs=1000, verbose=0)

model.predict(np.array([[9]]))
```
