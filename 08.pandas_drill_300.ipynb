# Series 생성 및 활용

```python
sr = pd.Series([100,200,300])
sr.index = ['철수', '영희', '아름']
sr.to_dict()
```

```python
sdic = {'철수':100, '영희':200, '아름':300}
sr = pd.Series(sdic)
sr.index, sr.values

sr.to_excel('./data/test.xlsx', header=False) # header = True/False

sr2 = pd.read_excel('./data/test.xlsx', index_col=0, header=None)
sr2
```
