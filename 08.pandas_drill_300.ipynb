# Series 생성 및 활용

```python
sr = pd.Series([100,200,300])
sr.index = ['철수', '영희', '아름']
sr.to_dict()
```

```python
sdic = {'철수':100, '영희':200, '아름':300}
sr = pd.Series(sdic)
sr.index, sr.values
sr
```

```python
sr.dtypes, sr.index.dtype
```

```python
sr.to_excel('./data/test.xlsx', header=False) # header = True/False

df = pd.read_excel('./data/test.xlsx', index_col=0, header=None)
# df = pd.read_excel('./data/test.xlsx', index_col=0, header=None).iloc[:,0]

df.index.name = 'name'
df.columns = ['score']

print("TYPES :", df.dtypes, type(df), sep='\n----------------------------\n')
df
```
