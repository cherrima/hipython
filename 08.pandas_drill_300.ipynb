```python
# BTC 금액이 50000 달러, XRP 금액이 1 달러, ETH 금액이 4000 달러일 때 시리즈 객체로 생성하세요. 시리즈 객체를 생성할 때 가격의 데이터 타입을 float로 지정해서 생성하세요.
# s = pd.Series({'BTC':50000, 'XRP':1, 'ETH':4000}).astype('float')

tickers = ["BTC", "XRP", "ETH"]
prices = [50000, 1, 4000]

s = pd.Series(data=prices, index=tickers, dtype='float')
s
```

```python
data = [100, 300, 200]
index = ["08/01", "08/02", "08/03"]
s = pd.Series(data, index)

s.sort_values(ascending=True)
s.sort_index(ascending=False)

print('*'*50)
print(s.iloc[ [0, 2, 1]])
print(s.loc[ ["08/01", "08/02", "08/03"]])
```
