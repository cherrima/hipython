```python
import anthropic
import os

# API 키 설정 (환경 변수 또는 직접 입력)
api_key = os.environ.get("ANTHROPIC_API_KEY")  # 또는 직접 입력: "your-api-key-here"

# Anthropic 클라이언트 초기화
client = anthropic.Anthropic(api_key=api_key)

# 기본 메시지 전송 예제
def basic_message():
    """간단한 메시지 전송"""
    message = client.messages.create(
        model="claude-sonnet-4-5-20250929",
        max_tokens=1000,
        messages=[
            {"role": "user", "content": "안녕하세요! Python에 대해 간단히 설명해주세요."}
        ]
    )
    print("응답:", message.content[0].text)
    return message

# 대화 이력을 유지하는 예제
def conversation_example():
    """여러 턴의 대화 진행"""
    conversation_history = [
        {"role": "user", "content": "Python의 장점은 무엇인가요?"},
    ]
    
    # 첫 번째 응답
    message = client.messages.create(
        model="claude-sonnet-4-5-20250929",
        max_tokens=1000,
        messages=conversation_history
    )
    
    assistant_response = message.content[0].text
    print("Claude:", assistant_response)
    
    # 대화 이력에 추가
    conversation_history.append({"role": "assistant", "content": assistant_response})
    conversation_history.append({"role": "user", "content": "그렇다면 단점은요?"})
    
    # 두 번째 응답
    message = client.messages.create(
        model="claude-sonnet-4-5-20250929",
        max_tokens=1000,
        messages=conversation_history
    )
    
    print("Claude:", message.content[0].text)
    return conversation_history

# 스트리밍 응답 예제
def streaming_example():
    """실시간으로 응답을 스트리밍"""
    print("스트리밍 응답:")
    
    with client.messages.stream(
        model="claude-sonnet-4-5-20250929",
        max_tokens=1000,
        messages=[
            {"role": "user", "content": "짧은 시를 하나 지어주세요."}
        ]
    ) as stream:
        for text in stream.text_stream:
            print(text, end="", flush=True)
    
    print("\n")

# 시스템 프롬프트 사용 예제
def system_prompt_example():
    """시스템 프롬프트로 역할 지정"""
    message = client.messages.create(
        model="claude-sonnet-4-5-20250929",
        max_tokens=1000,
        system="당신은 친절한 Python 프로그래밍 튜터입니다. 초보자도 이해하기 쉽게 설명해주세요.",
        messages=[
            {"role": "user", "content": "리스트 컴프리헨션이 뭔가요?"}
        ]
    )
    print("응답:", message.content[0].text)
    return message

# 에러 처리 예제
def error_handling_example():
    """API 호출 시 에러 처리"""
    try:
        message = client.messages.create(
            model="claude-sonnet-4-5-20250929",
            max_tokens=1000,
            messages=[
                {"role": "user", "content": "Hello!"}
            ]
        )
        print("성공:", message.content[0].text)
        
    except anthropic.APIError as e:
        print(f"API 에러 발생: {e}")
    except Exception as e:
        print(f"예상치 못한 에러: {e}")

# 메인 실행
if __name__ == "__main__":
    print("=== Claude API 사용 예제 ===\n")
    
    # 사용하고 싶은 예제의 주석을 해제하세요
    basic_message()
    # conversation_example()
    # streaming_example()
    # system_prompt_example()
    # error_handling_example()
```
