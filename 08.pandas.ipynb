# loc vs iloc
- DataFrame.loc['행이름':'행이름', '열이름':'열이름']
- DataFrame.iloc['행번호':'행번호', '열번호':'열번호']]

```python
# 연습문제
# [기본]

# 모든 행에서 이름 열만 선택하세요.
print(df['이름'])

# 모든 행에서 이름과 score 열을 선택하세요.
print(df[['이름', '점수']])

# 인덱스 C 행의 모든 데이터를 선택하세요.
print(df.loc['C'])

# 인덱스 D 행의 age 값만 선택하세요.
print(df.loc['D','나이'])
```

```python
# [조건 + loc]

# age가 27 이상인 행만 선택하세요.
print(df.loc[df['나이']>27])

# score가 90 이상인 사람의 이름과 score만 출력하세요.
print(df.loc[df['점수']>=90, ['이름', '점수']])

# age가 25 이상이면서 score가 90 이상인 행을 모두 선택하세요.
print(df.loc[(df['점수']>=90) & (df['나이']>27)])

# iloc 인덱싱 & 슬라이싱 연습문제
```

```python
# [슬라이싱]

# 인덱스 B부터 E까지의 모든 행을 선택하세요.
print(df.loc['B':'E'])

# 인덱스 C부터 F까지의 이름과 age 열만 선택하세요.
print(df.loc['B':'E', ['이름', '나이']])

# 인덱스 A부터 G까지에서 score 열만 선택하세요.
print(df.loc['A':'G', ['점수']])
```

```python
# 👉 정수 위치 기반 (0부터 시작)

# [기본]

# 첫 번째 행의 모든 데이터를 선택하세요.
print(df.iloc[0])

# 세 번째 행의 이름 값만 선택하세요.
print(df.iloc[2, 0])

# 처음 3개의 행을 선택하세요.
print(df.head(3))

# 마지막 행의 score 값만 선택하세요.
print(df.iloc[-1, 2])
```

```python
# [슬라이싱]

# 1번 행부터 4번 행까지 선택하세요.
print(df.iloc[1:5])

# 2번 행부터 5번 행까지에서 이름과 age 열만 선택하세요.
print(df.iloc[2:6, 0:2])

# 모든 행에서 age 열만 선택하세요.
print(df.iloc[:, 2])
```

```python
# [조합 문제]

# score 열만 선택하되, 앞에서 4개의 행만 출력하세요.
# print(df.head(4)['점수'])
print(df['점수'].head(4))

# 이름과 score 열을 선택하되, 마지막 3개의 행만 출력하세요.
print(df[['이름','점수']].tail(3))

# 짝수 번째 행(0, 2, 4, 6)의 모든 데이터를 선택하세요.
print(df.iloc[0::2])
```

```python
import pandas as pd 

data = {
    "name": ["김민수", "이서연", "박현우", "최지은", "정도윤"],
    "age": [29, 34, 41, 27, 36],
    "salary": [52000, 68000, 75000, 48000, 82000]
}

# DataFrame 생성
df = pd.DataFrame(data)

df.info()

def age_category(age) :
    age_cat = ""
    if age < 30 : 
        age_cat = "20대이하"
    elif age < 40 :
        age_cat = "30대"
    elif age < 50 :
        age_cat = "40대"
    else :
        age_cat = "50대이상"
    
    return age_cat
        

df["age category"] = df['age'].apply(age_category)
df['bonus'] = [salary*0.2 for salary in df['salary']]
df['bonus'] = df['bonus'].astype(int)
df.info
```
