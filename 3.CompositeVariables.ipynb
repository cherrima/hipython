```python
# dir (list)
import copy

score = [10, 20, 30, 40, 50]
shallow_score = score   # 이름만 다른 동일한 리스트
new_score = score[:]
deep_score = copy.copy(score) # 별도의 리스트 복사본

shallow_score[0]= 100
score[3]= -10
deep_score[1] = 99

new_score[4] = 888

print(id(score), score)
print(id(shallow_score), shallow_score)
print(id(new_score), new_score)
print(id(deep_score), deep_score)
```

```python
# list with dict
from operator import itemgetter

students = [{'num':1, 'name':'Kim', 'kor':80, 'math':90, 'eng':100},
            {'num':2, 'name':'Park', 'kor':70, 'math':50, 'eng':81},
            {'num':3, 'name':'Ghang', 'kor':99, 'math':99, 'eng':100},
            {'num':4, 'name':'Hong', 'kor':45, 'math':78, 'eng':90},
            {'num':5, 'name':'Yun', 'kor':78, 'math':90, 'eng':81}
           ]

for student in students :
    student['total'] = student['kor'] + student['math'] + student['eng']
    student['average'] = round(student['total'] / 3, 2)
    totals.append(student['total'])
    print(student)
    
print("-"*100)

sorted_students = sorted(students, key=itemgetter('total'), reverse=True)

rank = 0
for student in sorted_students :
    rank += 1
    student['rank'] = rank
    print(student)

print("-"*100)

students = sorted(students, key=itemgetter('num'))
for student in students :
        print(student)
```

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print(A | B)  # {1, 2, 3, 4, 5, 6} (합집합)
print(A & B)  # {3, 4} (교집합)
print(A - B)  # {1, 2} (차집합)
print(A ^ B)  # {1, 2, 5, 6} (XOR)

for item in A :
    print(item) 

# dir(A)
print(A.union(B))                    # (A | B) 
print(A.intersection(B))             # (A & B)  
print(A.difference(B))               # (A - B)
print(A.symmetric_difference(B))     # (A ^ B)
```
