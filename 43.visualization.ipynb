```python
import seaborn as sns

anscombe = sns.load_dataset('anscombe')
anscombe
```

```python
from matplotlib import pyplot as plt

plt.title('클래스와 오브젝트')
plt.xlabel('클래스')
plt.ylabel('오브젝트')
```

```python
d1 = anscombe[anscombe['dataset']=='I']
d2 = anscombe[anscombe['dataset']=='II']
d3 = anscombe[anscombe['dataset']=='III']
d4 = anscombe[anscombe['dataset']=='IV']
```

```python
# plt.figure(figsize=(10,6))

from matplotlib import pyplot as plt


fig = plt.figure(figsize=(10,6))
axes1 = fig.add_subplot(2,2,1) #2row, 2col,position
axes2 = fig.add_subplot(2,2,2)
axes3 = fig.add_subplot(2,2,3)
axes4 = fig.add_subplot(2,2,4)

axes1.plot( d1['x'], d1['y'] , 'o')
axes2.plot( d2['x'], d2['y'] , 'o')
axes3.plot( d3['x'], d3['y'] , 'o')
axes4.plot( d4['x'], d4['y'] , 'o')

axes1.set_title('dataset I')
axes2.set_title('dataset II')
axes3.set_title('dataset III')
axes4.set_title('dataset IV')

fig.suptitle('Anscombe Dataset 으로 보는 데이터 시각화의 의미')
fig.set_tight_layout(True)
plt.show()
```

# 용도에 맞는 차트 그리기

```python
tips_df = sns.load_dataset('tips')
tips_df
```

```python
# Total Bill vs Tip

plt.subplot(1,2,1)
plt.hist(data=tips_df, x='total_bill', bins=10, rwidth=0.8)


plt.subplot(1,2,2)
plt.hist(data=tips_df, x='tp', bins=10, rwidth=0.8)

plt.grid()
plt.show()
```

```python
plt.boxplot(
    [tips_df['total_bill'], tips_df['tip']],
    tick_labels=['total_bill', 'tip']
)

plt.grid()
plt.show()
```

```python
tips_df[tips_df['sex'] == 'Female'] ['total_bill']
tips_df[tips_df['sex'] == 'Female'] ['tip']
```

```python
plt.boxplot(
    [tips_df[tips_df['sex'] == 'Female'] ['total_bill'], tips_df[tips_df['sex'] == 'Female']['tip']],
    tick_labels=['total_bill(Female)', 'tip(Female)']
)

plt.grid()
plt.show()
```

```python
plt.boxplot(
    [tips_df[tips_df['sex'] == 'Male'] ['total_bill'], tips_df[tips_df['sex'] == 'Male']['tip']],
    tick_labels=['total_bill(Male)', 'tip(Male)']
)

plt.grid()
plt.show()
```

```python
tips_df['day'].unique()
```

```python
plt.figure(figsize=(12,6))
plt.boxplot(
    [tips_df[tips_df['day'] == 'Sun'] ['total_bill'], tips_df[tips_df['day'] == 'Sun']['tip'],
        tips_df[tips_df['day'] == 'Sat'] ['total_bill'], tips_df[tips_df['day'] == 'Sat']['tip'],
        tips_df[tips_df['day'] == 'Fri'] ['total_bill'], tips_df[tips_df['day'] == 'Fri']['tip'],
        tips_df[tips_df['day'] == 'Thur'] ['total_bill'], tips_df[tips_df['day'] == 'Thur']['tip']],
    tick_labels=['total_bill(Sun)', 'tip(Sun)', 'total_bill(Sat)', 'tip(Sat)',
                 'total_bill(Fri)', 'tip(Fri)', 'total_bill(Thur)', 'tip(Thur)']
)

plt.grid()
plt.show()
```

```python
# 요일별 테이블 매출 및 팁
plt.figure(figsize=(10,6))
plt.boxplot(
    [tips_df[tips_df['day'] == 'Sun']['tip'],
        tips_df[tips_df['day'] == 'Sat']['tip'],
        tips_df[tips_df['day'] == 'Fri']['tip'],
        tips_df[tips_df['day'] == 'Thur']['tip']],
    tick_labels=['tip(Sun)',  'tip(Sat)',
                 'tip(Fri)',  'tip(Thur)']
)

plt.grid()
plt.show()
```

```python
# 요일-시간 대별 방문횟수
gr_sr = tips_df.groupby(['day', 'time'], observed=False).size() # 그룹별 레코드 카운트
gr_sr.index = [f"{day}.{time}" for day, time in gr_sr.index]
gr_sr.plot(kind='bar', color='skyblue', figsize=(10, 6))

plt.title('Day & Time Counts')
plt.xlabel('Day.Time')
plt.ylabel('Count')
plt.xticks(rotation=45) # 글자가 겹치지 않게 회전
plt.show()
```
