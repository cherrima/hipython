# Class

- 속성(Attribute)
    * self.속성명 : 인스턴스 속성
    * 속성명 : 클래스 속성

- 메소드(Method)
    * 생성자 : __init__(self, args) : 생성자, 매개변수 => 인스텀스 속성 설장
    * 메소드 : def method(self, args) : 인스턴스명.method(args) 로 호출

- Method Decorator (@)
    * @classmethd : Class Method로 지정
        - def class_method(cls) : 인스턴스 self 대산 Class 변수 cls를 이용

    * @staticmethod : Class 변수 및 인스턴스 변수를 사용하지 않는 함수
        - def static_method(args) : self, cls 사용 못함

```python
class Car :
    # Class 공통속성
    instance_cnt = 0 
        
    def __init__(self, color='빨강'):
        self.color = color
        Car.instance_cnt += 1
        print(f"{self.color} 자동차가 생성되었습니다.") 
        
    
    def move(self):
        print(f"{self.color} 자동차 움직입니다.") 
       
    def self_drive(self) :
        print(f"자율주행으로 {Car.instance_cnt}대가 운행합니다.")
    
    @classmethod   
    def check_count(cls) :
        print(f"차량 대수 : {cls.instance_cnt}")
        
    @staticmethod
    def check_type(code) : #static method
        if code > 20 :
            print('전기자동차')
        elif 10 < code < 20 :
            print('가솔린자동차')
        else:
            print('디젤차')
        
    # def whohas(self):
    #    print(f"이 {self.color} 자전거는 {self.owner} 소유입니다") 
```

```python
car = Car()
car.move()
car.self_drive()

car_yellow = Car('Yellow')
car_yellow.move()
car_yellow.self_drive()

car.check_count()
Car.check_count()

car.check_type(25)
```

```python
class Calculator :
    
    def __init__(self, num_list) :
        self.num_list = num_list
    
    def add(self) : 
        result = 0
        for num in self.num_list :
            result += num
        return result
    
    def mul(self) :
        result = 1
        for num in self.num_list :
            result *= num
        return result
    
    def avg(self) :
        return self.add() / len(self.num_list)
    
    def max(self) :
        max_val = self.num_list[0]
        for num in self.num_list :
            if num > max_val :
                max_val = num
        return max_val
```

```python
cal = Calculator([1,2,3,4,5])
print(cal.add()) 
print(cal.mul())
print(cal.avg())
print(cal.max())
```

```python
# Account 클래스 작성
# 속성 : 은행이름(bank_name, 국민은행), 계좌번호(account_number), 예금주(owner), 잔액(balance, 0)
# 메서드 : 입금(deposit), 출금(withdraw), 잔액조회(check_balance)   

# This Code Segment is first genrated by copilot

import random

class Account :
    bank_name = "국민은행"
    
    def __init__(self, owner, balance=0) :
        random.seed(42) 
        self.account_number = str(random.randint(1000000000, 9999999999))
        self.owner = owner
        self.balance = balance
        
        str_account_number = self.account_number[:3] + '-' + self.account_number[3:5] + '-' + self.account_number[5:]
        print(f"{self.owner}님의 계좌가 생성되었습니다. 계좌번호 : {str_account_number}, 잔액 : {self.balance}원")
    
    def deposit(self, amount) :
        self.balance += amount
        print(f"{amount}원이 입금되었습니다. 잔액은 {self.balance}원입니다.")
    
    def withdraw(self, amount) :
        if amount > self.balance :
            print("잔액이 부족합니다.")
        else :
            self.balance -= amount
            print(f"{amount}원이 출금되었습니다. 잔액은 {self.balance}원입니다.")
    
    def check_balance(self) :
        print(f"{self.owner}님의 잔액은 {self.balance}원입니다.")
```

```python
acc = Account("강창희", 10000)  

acc.deposit(5000)
acc.withdraw(3000)  
acc.withdraw(15000)  
acc.check_balance()

```
