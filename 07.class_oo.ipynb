```python
class Bicycle :
    def __init__(self, owner = '강창희', color='빨강'):
        self.owner = owner
        self.color = color
    
    def move(self):
       print(f"{self.color} 자전거가 움직입니다") 
    
    def whohas(self):
       print(f"이 {self.color} 자전거는 {self.owner} 소유입니다") 
```

```python
b1 = Bicycle(color="노랑")
b2 = Bicycle(owner='홍길동')

print(type(b1))
b1.move()
b2.move()

b1.whohas()
b2.whohas()
```

```python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"안녕하세요. 제 이름은 {self.name}이고, 나이는 {self.age}살입니다.")
```

```python
stu = Student("강창희", 56)
stu.introduce()
```
