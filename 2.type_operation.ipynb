```python
# 사칙연산자, 비교 연산자, 논리연산자, 대입연산자, etc
# 사칙 : + - * / // %
# 비교 : > < >= <= == != is
# 축약 : += -+ *= /=
# 논리 : and or not & | !  in

# __add__   :  + 연산자 지원
print("a".__add__("b"), "a"+"b") #  a+b

# == vs is
# == 실제값 비교, is 주소값 비
lunch_1 = ['주먹밥', '치킨', '칼국수']
lunch_2 = ['주먹밥', '치킨', '칼국수']

print(lunch_1 == lunch_2, lunch_1 is lunch_2)
print(id(lunch_1) == id(lunch_2), lunch_1 is lunch_2)
```
