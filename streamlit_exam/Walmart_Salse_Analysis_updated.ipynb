# Wallmart 데이터 분석

## 프로젝트 개요
[Kaggle의 Wallmart 데이터 셋](https://www.kaggle.com/datasets/harshalpanchal/walmart-black-friday-sales) 분석 프로젝트입니다. 데이터 전처리, 탐색적 데이터 분석(EDA)을 통한 데이터의 분포를 확인하고, VIP 고객 도출 및 지역별 구매 고객 특징을 파악하여 마케팅 전략을 도출합니다.
이 프로젝트의 목표는 주어진 데이터의 분포를 분석하여 주요 특징을 도출하고, 이를 바탕으로 마케팅 전략을 제시하는 것입니다.

## 데이터 설명
- 데이터 출처: [Kaggle의 Wallmart 데이터 셋](https://www.kaggle.com/datasets/harshalpanchal/walmart-black-friday-sales) 
- 데이터 상세: 약 55만개의 사용자 거래 내역과 10개의 특징(550068, 10)
  - `User_ID`: User ID
  - `Product_ID`: Product ID
  - `Gender`: Sex of User
  - `Age`: Age in bins
  - `Occupation`: Occupation(Masked)
  - `City_Category`: Category of the City (A,B,C)
  - `StayInCurrentCityYears`: Number of years stay in current city
  - `Marital_Status`: Marital Status
  - `ProductCategory`: Product Category (Masked)
  - `Purchase`: Purchase Amount

## 제품의 수 및 카테고리별 제품의 수

```python
# 제품 종류 수 및 카테고리별 집계
product_cnt = df['Product_ID'].nunique()
print("제품 종류 수(Product_ID unique):", product_cnt)

product_cat_df = df.groupby('Product_Category').agg(
    Product_Kinds=('Product_ID', 'nunique'),
    Total_Purchase=('Purchase', 'sum')
)

# 제품당 평균구매액 = 구매액(합) / 제품수
product_cat_df['Avg_Purchase_per_Product'] = (product_cat_df['Total_Purchase'] / product_cat_df['Product_Kinds']).round(0)

# 구매액 기준 정렬
product_cat_df = product_cat_df.sort_values('Total_Purchase', ascending=False)
product_cat_df
```

제품 종류는 **3631개** - 카테고리별 **2개**에서 많게는 **1047개** 임<br>
카테고리의 제품이 많을 수록 카테고리 구매액도 증가 (**상관도 0.724**)


```python
# 카테고리별 제품수
product_cat_df['Product_Kinds'].plot(kind='bar', figsize=(10, 6))
plt.xticks(rotation=0)
plt.title('제품 카테고리별 제품수')
plt.show()
```

```python
# 카테고리별 구매액(합)
product_cat_df['Total_Purchase'].plot(kind='bar', figsize=(10, 6))
plt.gca().yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,}'))
plt.xticks(rotation=0)
plt.title('제품 카테고리별 총구매액(합)')
plt.show()
```

```python
# 제품당 평균구매액(구매액/제품수)
product_cat_df['Avg_Purchase_per_Product'].sort_values(ascending=False).plot(kind='bar', figsize=(10, 6))
plt.gca().yaxis.set_major_formatter(ticker.StrMethodFormatter('{x:,}'))
plt.xticks(rotation=0)
plt.title('제품당 평균구매액(구매액/제품수)')
plt.show()
```

```python
# 카테고리별 제품수 / 구매액 / 제품당 평균구매액 상관관계
product_cat_df.corr(numeric_only=True)
```

```python
import numpy as np
```

```python

# 2x2 서브플롯 생성 (figsize로 크기 조절)
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# 고가품 10, 7, 6, 9
high_categories = [10, 7, 6, 9]

stat_cat_age = (df.groupby(['Age']).size() / df.shape[0]).to_frame()
stat_cat_age.columns = ['All Categoires']

x = np.arange(len(stat_cat_age.index))  # x축 눈금 위치 (0, 1, 2...)
width = 0.35  

for i in range(0, 4) :
    ix, iy = divmod(i,2)
    cat = high_categories[i]
    
    cat_column =  f"Category {cat}"
    df_catN = df[df['Product_Category']==cat]
    stat_cat_age[cat_column] = df_catN.groupby(['Age']).size() / df_catN.shape[0]
    
    # axs[ix, iy].plot(stat_cat_age.index, stat_cat_age[['All Categoires', cat_column]])
    # axs[ix, iy].set_title(f"Category {cat}")
    
    # 첫 번째 막대 (왼쪽으로 width/2 이동)
    axs[ix, iy].bar(x - width/2, stat_cat_age['All Categoires'], width, label='All Categoires')
    
    # 두 번째 막대 (오른쪽으로 width/2 이동)
    axs[ix, iy].bar(x + width/2, stat_cat_age[cat_column], width, label=cat_column)
    
    axs[ix, iy].set_xticks(x)
    axs[ix, iy].set_xticklabels(stat_cat_age.index)
    axs[ix, iy].set_title(cat_column)
    axs[ix, iy].legend()


plt.show()
# for category in high_categories
```

```python
# 나이별 구매 비중 분포가 상이한 제품 카테고리 (10, 12, 17, 18)
fig, axs = plt.subplots(2, 2, figsize=(10, 8))

high_categories = [10, 12, 17, 18]

stat_cat_age = (df.groupby(['Age']).size() / df.shape[0]).to_frame()
stat_cat_age.columns = ['All Categories']

for i in range(0, 4):
    ix, iy = divmod(i, 2)
    cat = high_categories[i]

    cat_column = f"Category {cat}"
    df_catN = df[df['Product_Category'] == cat]
    stat_cat_age[cat_column] = df_catN.groupby(['Age']).size() / df_catN.shape[0]

    cols = ['All Categories', cat_column]
    axs[ix, iy].plot(stat_cat_age.index, stat_cat_age[cols], label=cols)
    axs[ix, iy].set_title(cat_column)
    axs[ix, iy].legend()

plt.tight_layout()
plt.show()

stat_cat_age
```

```python
# stat_cat_age.describe() 중 평균(mean)과 표준편차(std)
stat_cat_age[['All Categories'] + [f'Category {c}' for c in high_categories]].describe().loc[['mean','std']]
```

```python
categories = df['Product_Category'].unique()
categories.sort()
categories

stat_cat_age = (df.groupby(['Age']).size() / df.shape[0]).to_frame()
stat_cat_age.columns = ['All Categoires']

for cat in categories :
    cat_column =  f"Category {cat}"
    df_catN = df[df['Product_Category']==cat]
    stat_cat_age[cat_column] = df_catN.groupby(['Age']).size() / df_catN.shape[0]
    
stat_cat_age

stat_cat_age.iloc[:, 0:6].plot(kind='line', figsize=(12, 6))
# stat_cat_age.plot(kind='line', figsize=(16, 10))
plt.xticks(rotation=0) 
plt.ylabel("Purchase Percentage") 


stat_cat_age.iloc[:, [0, 6,7,8,9,10]].plot(kind='line', figsize=(12, 6))
# stat_cat_age.plot(kind='line', figsize=(16, 10))
plt.xticks(rotation=0) 
plt.ylabel("Purchase Percentage") 


stat_cat_age.iloc[:, [0,11,12,13,14,15]].plot(kind='line', figsize=(12, 6))
# stat_cat_age.plot(kind='line', figsize=(16, 10))
plt.xticks(rotation=0) 
plt.ylabel("Purchase Percentage") 

stat_cat_age.iloc[:, [0,16,17,18,19,20]].plot(kind='line', figsize=(12, 6))
# stat_cat_age.plot(kind='line', figsize=(16, 10))
plt.xticks(rotation=0) 
plt.ylabel("Purchase Percentage")

stat_cat_age.plot(kind='line', figsize=(15, 7))
# stat_cat_age.plot(kind='line', figsize=(16, 10))
plt.xticks(rotation=0) 
plt.ylabel("Purchase Percentage")

plt.show()
    
```
