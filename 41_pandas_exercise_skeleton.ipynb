## Q24.
<div class="alert alert-success">
[문제24]     
    
smoker 여부에 따라, group_val 컬럼에 '흡연자', '비흡연자' 표시하기
</div>

```python
df['smoker'].map({'yes':'흡연자', 'no':'비흡연자'})
```

```python
df = df.drop(columns=['group_val'])

smoker_region_df.indices.keys()

```

```python
def issmoker(smoker_yn) :
    return '흡연자' if smoker_yn=='yes' else '비흡연자'

df['smoker'].map(issmoker)
```

```python
group_val = {'smoker':['yes', 'no'], 'smoker_group' : ['흡연자', '비흡연자']}

smoker_group = pd.DataFrame(group_val)
smoker_group
```

```python
smoke_val = {'yes':'흡연자', 'no':'비흡연자'}

smoke_sr = pd.Series(smoke_val, name='흡연자여부')
smoke_sr
```

```python
pd.merge(df, smoke_sr, left_on='smoker', right_index=True)
```

```python
pd.merge(df, smoker_group, on='smoker')
```
