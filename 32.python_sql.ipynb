```python
import pymysql

def get_connection() :
    try :
        con = pymysql.connect(
            host = 'localhost', # '127.0.0.1',
            user = 'root',
            password = '1234',
            database = 'wntrade',
            charset ='utf8'  # 한글처리
        )
        print("접속 OK")
        print("작업 후 커서 및 MySQL 서버 접속 종료하세요")


    except Exception as ex :
        print("접속 실패 :", ex)
        print("MySQL 서비스가 실행 중인지 확인 요망!!!")
        
    return con
```

```python
import pymysql

con = get_connection()

try :
    with con.cursor() as cursor :
        
        sql_query = ''' SELECT 고객회사명, 담당자명
                        FROM 고객 JOIN 주문 ON 주문.고객번호 = 고객.고객번호
                        WHERE 주문번호 = 'H0250' ; '''
                        
        sql_query_2 = '''SELECT 고객번호, 고객회사명, 담당자명, 도시, 마일리지
                        FROM 고객
                        WHERE (도시, 마일리지) IN (SELECT 도시, MAX(마일리지) FROM 고객 GROUP BY 도시) ; '''

        # cursor.execute(sql_query)
        cursor.execute(sql_query_2)
        # 쿼리 결과 가져오기
        result = cursor.fetchall()
        # 결과 출력
        print('-'*80)
        for row in result:
            print(row)
            
        print('*'*80)

except Exception as ex :
    print("Query 실행 오류 :", ex)
    
finally :
    cursor.close()
    con.close()
    print("MySQL 서버 접송 종료!!!")
```

```python
import pymysql

con = get_connection()

try :
    with con.cursor() as cursor :                     
        sql_query = '''SELECT 고객번호, 고객회사명, 담당자명, 도시, 마일리지
                        FROM 고객
                        WHERE (도시, 마일리지) IN (SELECT 도시, MAX(마일리지) FROM 고객 GROUP BY 도시) ; '''

        cursor.execute(sql_query)
        
        print("Retrieval Count :", cursor.rowcount)
        
        if ( cursor.rowcount > 0 ) :
            print('-'*80)
            while True :
                rows_10 = cursor.fetchmany(10)
                if rows_10 is None : break 
                print("Fetched :", rows_10.count)
                for row in rows_10 :
                    print(row)
            print('*'*80)

except Exception as ex :
    print("Query 실행 오류 :", ex)
    
finally :
    cursor.close()
    con.close()
    print("MySQL 서버 접송 종료!!!")
```
