```python
import pymysql

try :
    con = pymysql.connect(
        host = 'localhost', # '127.0.0.1',
        user = 'root',
        password = '1234',
        db = 'wntrade',
        charset ='utf8'
    )
    cursor = con.cursor()

except Exception as ex :
    print("접속 실패 :", ex)
```

```python
# 쿼리 실행 예시
sql_query = ''' select 사원번호, 이름 AS 직원명, 직위, 입사일
                from 사원
                order by 사원번호 asc ; '''

cursor.execute(sql_query)
# 쿼리 결과 가져오기
result = cursor.fetchall()
# 결과 출력
for row in result:
 print(row)
```

```python
cursor.close()
con.close()
```
