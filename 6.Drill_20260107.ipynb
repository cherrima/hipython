```python
# Q1. 피보나치 수열

def fibonacci_recursive(number) :
    if ( number <= 0 ) : return 0
    if ( number == 1 ) : return 1

    return fibonacci(number-1) + fibonacci(number-2)

def fibonacci(number) :
    if ( number <= 0 ) : return 0
    if ( number == 1 ) : return 1

    fibo_list = [0, 1]

    for num in range(2, number+1) :
        fibo =  fibo_list[num-1] + fibo_list[num-2]
        fibo_list.append(fibo)

    return fibo_list[-1]

number = 9
#print("Fibonacci number for", number-2, "is", fibonacci(number-2))
#print("Fibonacci number for", number-1, "is", fibonacci(number-1))
print("Fibonacci number for", number, "is", fibonacci(number))
```

```python
# Q2. 문자열을 입력 받아 같은 문자가 연속적으로 반복되는 경우에 그 반복 횟수를 표시해 문자열을 압축하여 표시해 보자
instr = "aaabbcccccaa"

def str_zip(text) :

    if len(text) == 0 : return ""
    
    result_str =""
    prev_char = text[0]
    repeat = 1

    for char in text[1:] :
        if char == prev_char :
            repeat += 1
        else :
            result_str += f"{prev_char}{repeat}"
            prev_char = char
            repeat = 1

    return result_str + f"{prev_char}{repeat}"
    
print(str_zip(instr))
```

```python
# Q3. 0 ~ 9 까지의 문자로 된 숫자를 입력받았을 때, 이 입력값이 0 ~ 9까지의 모든 숫자가 각각 한 번씩만 사용된 것인지 확인하는 함수
# 입력예시 : 0123456789 01234 01234567890 6789012345 012322456789
# 출력예시 : true false false true false

def consumed_all(num_text) :

    TEST_SET = set("0123456789")

    if ( len(num_text) != 10 ) : return False 

    # return all(char in TEST_STR for char in num_text)

    return set(num_text) == TEST_SET

for str in ['0123456789','01234', '01234567890', '6789012345', '012322456789'] :
    print(str, consumed_all(str))
```

```python
# Q4. 콤마가 포함된 문자열 숫자를 입력받아 정수로 변환하는 convert_int 함수
def convert_int(numstr) :
    # return int("".join(numstr.split(',')))
    return int(numstr.replace(',', ''))

numstr = "1,234,567"

print(convert_int(numstr))
```
