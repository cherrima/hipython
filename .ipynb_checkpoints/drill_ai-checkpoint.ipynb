```python
from openai import OpenAI
from dotenv import load_dotenv
import os

# print(os.getenv("PATH"))

# .env 파일 로드
load_dotenv()

# 헬퍼 
def query_helper(text, role="user"):
    query = {
        "role": role,
        "content": [{"type":"input_text","text": text}]
    }
 
    return query

# Prompt 입력, 실행, 및 결과 리턴 API 키 읽기
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY")) # OpenAI 서버 연결
print(client)
```

```python
# system_prompt = """
#     역할 : B2B 마케팅 전문가로서 답변해줘
#     목표 : 인스타그램 랜덤 페이지에 넣을 핵심 문구 작성
#     조건 : 각 문구는 20자 이내로 작성, 신뢰성과 전문성 강조, 과장된 표햔과 감탄사 배제
#     출력형식 : 번호목록 형식 (1~5)
# """

# system_prompt = "당신은 마케팅 전문가야."

# user_prompt = """
#     "1인 컨설팅 기업에 대한 인스타그램 홍보 문구 만들어줘."
# """

system_prompt = """
     당신은 '깐깐하고 신경질적인' 상사야.
     보고서는 마감일 3일 전에 초안을 받아 검토해야 직성이 풀린다.
     가능하면 미완성된 초안이라도 중간에 검토하는 것을 선호한다.
"""

user_prompt = """
    A 직원에게 보고서 초안을 요청하는 메일을 경고성 문구를 포함한 강한 어조로 작성해줘
    경고성 문구는 해당 직원에 대한 직접적인 영향을 미치는 것으로 인사 평가, 중요 프로젝트 제외 등을 포함해 줘
    상황 : A 직원은 항상 마감일 직전에 보고서를 제출해서, 해당 직원에게 예민해진 상태야
"""

response = client.responses.create(
    model="gpt-4o-mini",
    input=[
         {"role" : "system", "content" : system_prompt},
         {"role" : "user", "content" : user_prompt} 
    ],
    temperature=1.0  # 응답 생성 시 무작위성(randomness) 또는 창의성(creativity)을 조절
    # top_p=tp
    # requency_penalty, presence_penalty 
)

def format_gpt_output(text):
    """
    GPT 출력에서 \n을 실제 줄바꿈으로 변환하여 가독성 있게 출력
    """
    # \n을 실제 줄바꿈으로 변환
    formatted_text = text.replace('\\n', '\n')
    return formatted_text
    
print("시스템 프롬프트 :")
print(system_prompt)
print("사용자 프롬프트 :")
print(user_prompt)
print("결과 :")
print(response.output_text)
```

```python
user_prompt = "A 직원에게 보고서 초안을 요청하는 메일을 작성하라.\n" \
             + "상황 : A 직원은 항상 마감일 직전에 보고서를 제출하고 있어"

characters = ['유머러스한', '깐깐한', '꼰대같은', '원칙주의적인', '급한 성격의']

for character in characters :
    
    system_prompt = f"당신은 {character} 상사야.\n 간결하고 직설적인 말투를 선호해."
                # + "공격적이지 않으면서 설득력있는 성격이야.\n" \
                
    
    response = client.responses.create (
        model="gpt-4o-mini",
        input=[
            {"role" : "system", "content" : system_prompt},
             {"role" : "user", "content" : user_prompt} 
        ],
        temperature=0.5  # 응답 생성 시 무작위성(randomness) 또는 창의성(creativity)을 조절
        # top_p=tp
        # requency_penalty, presence_penalty
    )
    
    print(f"Character {character}")
    print(response.output_text)
    print("-"*100)
```
