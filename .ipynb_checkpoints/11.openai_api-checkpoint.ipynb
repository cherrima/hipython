````python
from openai import OpenAI
from dotenv import load_dotenv
import os

# print(os.getenv("PATH"))

# .env 파일 로드
load_dotenv()

# 헬퍼 
def query_helper(role, text):
    return {
       "role": role,
        "content": [{"type":"input_text","text": text}]
    }
````

````python
# 환경변수에서 API 키 읽기
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

response = client.responses.create(
    model="gpt-4o-mini",
    input=[
        query_msg("system","너는 광고 전문가다."),
        query_msg("user","1인 컨설팅 기업에 대한 광고 문구를 영어로 만들어줘.")

        # {"role" : "system", "content" : "너는 광고 전문가다."},
        # {"role" : "user", "content" : "1인 컨설팅 기업에 대한 광고 문구 만들어줘."}
        
    ],
    temperature=0  # 응답 생성 시 무작위성(randomness) 또는 창의성(creativity)을 조절
)

# temperature 낮은 값 (0에 가까울수록): 모델이 더 결정적이고(deterministic) 일관된 응답
# 높은 값 (2.0에 가까울수록): 모델이 더 무작위적이고 창의적인 응답을 생성

print(response.output_text)
````

Sovereign AI refers to artificial intelligence systems that are designed to operate independently and autonomously, often with a focus on maintaining control over their own operations and decision-making processes. This concept can encompass various aspects, including:

1. **Autonomy**: Sovereign AI systems can make decisions without human intervention, allowing them to function in complex environments.

2. **Self-governance**: These systems may have built-in protocols to manage their own behavior and ensure compliance with ethical guidelines or regulations.

3. **Data Sovereignty**: This aspect emphasizes the control over data, ensuring that data used by AI systems is managed according to local laws and regulations.

4. **Security and Privacy**: Sovereign AI often incorporates measures to protect sensitive information and maintain user privacy.

Overall, the idea of Sovereign AI raises important discussions about ethics, governance, and the implications of autonomous systems in society.

````python
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

system_prompt = """
    역할 : 시사 전문가로서 답변해줘
    목표 : 입력된 뉴스를 분석하여 제시된 출력 형식으로 출력
    출력형식 : 
         1. 주제 : 1줄 이내
         2. 기사 요약 
            500자 또는 5줄 이내 목록 형식으로
         3. 가능한 질문 3가지
         * 키워드 : 파이선 리스트 형식으로
"""

user_content = """
네이버와 NC가 우리나라 고유 인공지능(AI) 모델을 뽑는 ‘국가대표 AI’ 1차 평가에서 탈락했다. 애초 정부는 다섯 팀 중 한 팀만 탈락시킨다는 계획이었지만 두 팀이 1차 경쟁에서 떨어진 것이다. 국가대표 AI 선발은 한국이 글로벌 AI 경쟁에서 뒤처지지 않도록 미·중에 이어 ‘AI 3강’이 되기 위해 독자적인 한국형 파운데이션 모델을 개발하고, 이를 뒷받침할 핵심 기업을 선정해 집중적으로 지원하는 국가 프로젝트다.

과학기술정보통신부는 15일 오후 정부서울청사에서 국가대표 AI 1차 평가 결과를 발표했다. 류제명 과기정통부 2차관은 “벤치마크(성능 평가 시험지) 평가(40점), 전문가 평가(35점), 사용자 평가(25점)를 통해 종합 평가한 결과 LG AI 연구원·SK텔레콤·업스테이지가 2단계에 진출했다”고 말했다. 벤치마크 평가, 전문가 평가, 사용자 평가에서 모두 LG AI연구원이 1위를 차지했다.


네이버클라우드는 AI 모델 성능 평가에서는 상위 4팀에 포함됐지만 탈락했다. 류 차관은 “네이버클라우드의 AI 모델은 독자성 기준을 충족하지 못하는 것으로 판단했고 전문가 평가위원들도 독자성 한계에 대한 문제를 제기했다”면서 “네이버클라우드는 독자 AI 파운데이션 모델 개발이라는 이번 프로젝트 조건에 부합하지 못하는 것으로 결정했다”고 말했다. NC AI는 AI 모델 성능 평가 점수가 낮아 탈락했지만 네이버클라우드는 독자 AI 모델을 개발하지 못해 탈락한 것이다. 정부는 탈락한 팀에 10일간 이의 제기 신청 기간을 부여하고 이의 제기가 없을 경우 이번 평가 결과가 확정된다고 밝혔다.

앞서 정부는 해외 기술에 의존하지 않는 독자적인 AI 파운데이션 모델을 개발해 AI 주권을 확보하겠다는 취지로 이번 국가대표 AI 프로젝트를 시작했다. 제미나이·챗GPT·딥시크 등 글로벌 프런티어급 AI 모델에 종속되지 않으려면 더 늦기 전에 독자 기술로 개발한 AI 모델을 확보해야 한다는 것이다. 15팀이 참여한 공모에서 지난해 8월 네이버클라우드·업스테이지·SK텔레콤·NC AI·LG AI연구원이 국대 AI로 선발됐다.

애초 정부는 1차 평가에서 1팀을 탈락시키고 이후 6개월마다 1팀씩 총 2팀을 떨어뜨려 이르면 올해 말 최종 2팀을 선발할 예정이었다. 그러나 계획과 달리 1차 평가에서 2팀이 탈락하면서 정부는 최대한 이른 시기에 공모를 통해 1팀을 추가로 선정해 2차 평가에 들어간다는 계획이다. 이번에 탈락한 네이버클라우드·NC AI를 포함해 모든 기업이 공모에 참여할 수 있다. 다만 처음 선발된 팀은 기존 팀과 달리 AI 모델 개발 기간이 짧아 동일 조건에서 경쟁하지 못하는 게 아니냐는 지적도 나온다. 네이버 관계자는 “재공모에 참여할 계획이 없다”고 말했다.


국내 AI 업계의 맏형 격인 네이버의 1차 평가 탈락 소식에 AI 업계에선 당혹스럽다는 반응이 나오고 있다. 애초 네이버클라우드는 LG AI 연구원과 함께 ‘국내 AI 2강’으로 꼽힐 만큼 앞선 기술력을 보유한 것으로 평가받았기 때문이다. AI 업계에서도 네이버가 1라운드에서 탈락할 것으로 보는 전문가는 거의 없었다. 다만 지난달 말 1차 평가를 앞두고 공개한 네이버의 AI 모델이 중국산 ‘비전 인코더’를 사용한 것으로 드러나 ‘프롬 스크래치(아무것도 없는 상태에서 개발했다는 뜻)’ 논란이 제기되면서 “네이버도 탈락 후보에 들어갈 수 있다”는 목소리가 커지기 시작했다.


네이버는 LLM(대형 언어 모델)을 개발한 나머지 4개 업체와 달리 이미지·영상 등을 인식하는 옴니모달 모델을 개발했다. 비전 인코더는 눈 역할을 한다. 특히 AI 전문가들은 네이버가 단순히 비전 인코더만 사용한 게 아니라 가중치(학습이 완료된 지능)를 재사용한 것으로 판단했다. 류 차관은 “독자 AI 파운데이션 모델은 가중치를 초기화한 후 다시 학습을 통해 가중치를 채워나가야 하는데, 네이버 기술 보고서에도 나왔지만, 네이버는 기존 가중치를 그대로 갖다 썼다”고 말했다.

AI 업계에선 이번 1차 평가로 과기정통부가 애초 목표로 했던 독자 AI 파운데이션 모델 개발 취지를 지킬 수 있게 됐다고 본다. 중국산 모듈을 사용한 네이버가 통과됐다면 이번 프로젝트의 근본 취지가 훼손될 수밖에 없었기 때문이다. 정부는 프롬 스크래치 기준도 제시했다. 류 차관은 “완전한 우리 기술로 AI 모델을 개발하거나, 라이선스 제약 없는 오픈소스를 활용해 스스로 개발해야 하며, 오픈소스 활용으로 인한 외부의 통제·간섭에서 자유로워야 한다”며 “기술적으로는 오픈소스를 쓰더라도 가중치를 초기화해 사용해야 한다”고 말했다. 오픈소스(Open Source)는 소프트웨어의 프로그램이 어떻게 작동하는지 보여주는 원본 코드를 공개해서 누구나 볼 수 있고, 필요하면 수정하거나 다시 배포할 수 있도록 한 소프트웨어 개발 방식이다.
"""

response = client.responses.create(
    model="gpt-4o-mini",
    input=[
        query_helper("system", system_prompt),
        query_helper("user",user_content),
    ],
    temperature=0
)

print(response.output_text)
````

````python
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

prompt = """주어진 고장내용 텍스트로부터 고장부품, 불량유형, 조치내용을 추출하시오.
결과는 반드시 파이썬 Dictionary 형태로만 반환하시오.

⚠️ 중요 규칙:
- 코드 블록(```)을 사용하지 마십시오.
- 설명 문장, 주석, 추가 텍스트를 절대 포함하지 마십시오.
- 오직 하나의 Python Dictionary 리터럴만 출력하십시오.

ex)
고장내용: '출하장 입구 에어 50A 유니온 에어리크 재조임'
출력: {"고장부품": "50A 유니온", "불량유형": "에어리크", "조치내용": "재조임"}

고장내용: 'SHIELD ROOM 화풍기 철거부위 마감,판넬벽체문타공'
출력: {"고장부품": "SHIELD ROOM 화풍기", "불량유형": "철거부위 마감", "조치내용": "판넬벽체문타공"}
"""

def ai_response(text, system_prompt = prompt) :
    response = client.responses.create(
        model="gpt-4o-mini",
        input=[
            query_helper("system", system_prompt),
            query_helper("user", "고장내용: " + text),
        ],
        temperature=0
    )
    return response.output_text

error_text = sub_df['고장내용'].loc[0] 
print(ai_response(error_text))

result = []
for error_text in sub_df['고장내용'] :
    output_text = ai_response(error_text)
    result.append(output_text)
    print(output_text)
````
