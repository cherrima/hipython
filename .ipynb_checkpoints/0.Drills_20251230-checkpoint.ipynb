```python
# 유저한테서 KMH값을 입력 받으면, 그에 맞는 MPH로 변환하는 코드를 만드세요. MPH = 0.6214 * KMH
# speed_kmh = float(input("속도를 km/h 단위로 입력하세요 :"))
M_VALUE = 0.6214

speed_kmh = 0
while speed_kmh <= 0 :
    temp = input("속도를 km/h 단위로 입력하세요 (양의 정수) :")
    print("[" + temp + "]", type(temp))
    speed_kmh = int(temp) if len(temp) > 0 else 0

print(f"입력한 속도는 {speed_kmh*M_VALUE:.2f}MPH입니다.")
```

```python
# 유저한테, 년도와 월을 입력받으세요. 그리고나서 해당 년도와 월을 출력하는 코드
import calendar

while True :
    year_month = input("연도 4자리와 월 2자리를 붙여서 입력하세요 (예: 202512) : ")
    if (len(year_month) > 5 ) :
        year = int(year_month[:4])
        month = int(year_month[4:6])
        if ( 0 < month <= 12 ) : break
        else : print("월은 12 이하 정수로 입력하세요")
    else :
        print("6자리 숫자로 입력하세요")
        
print(calendar.month(year, month))
```

```python
# 11. 다음 딕셔너리 students에서 name의 값을 리스트(name_list)로, age의 값을 튜플(age_tuple) 만드는 코드를 작성하세요
students = {
    "student1": {"name": "Alice", "age": 24},
    "student2": {"name": "Bob", "age": 22},
    "student3": {"name": "Charlie", "age": 23}
}

# name, age list
name_list = []
age_list = []
for student in students.values() :
    name_list.append(student["name"])
    age_list.append(student["age"])

age_tuple = tuple(age_list)
print(name_list)
print(age_tuple)
```
