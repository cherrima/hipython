# Comprehension(내포)

- 기본구조 : `[표현식 for 항목 in 반복가능객체 if 조건문]` <br>
  --> if 조건문은 선택사항
- 리스트, 딕셔너리, 집합 형태로 구성

```python
# Comprehension
# even_squares = [x*x for x in (2, 4, 6, 8)]
even_squares = [x*x for x in range(1, 10) if x%2 == 0 ]
print(even_squares)

# 주어진 크기의 원소를 가짐
zeros = [0 for x in range(1, 10) if x%2 == 0]
print(zeros)
zeros = [0 for _ in range(1, 10)]
print(zeros)
```

```python
# 두개의 변수를 받아서 +, * 연산 결과 반환
def add_multi(x, y) :
    return x+y, x*y

print(add_multi(10, 15))

# 파라미터 지정 - 키워드 파라미터
def introduce(name, message='지구를 지켜줍니다') :
    print(f"우리의  {name}님, {message}")

introduce('아이언맨', '젠틀해요')
introduce(name='어벤저스', message='Assemble!')


# 변수 미정 - Wild Card
def calc(a, b, *c) :
    sum_all = a + b + sum(c)
    return sum_all

print(calc(1, 2, 4, 6, 8))
```

# Snack Box

1. 함수 도출 : CRUD 기본
2. 함수 구현
3. 함수 문제점 도출 : 함수 리팩토링
4. 전체 흐름 구현 함수

```python
# 간식박스
def make_snackbox() :
    # snackbox = ['홈런볼', '두바이초콜릿']
    snackbox = []
    return snackbox

# 간식박스 추가함수
def add_snackbox(snackbox) :
    snacks = input("먹고 싶은 간식을 추가하세요. (쉼표로 분리) :").replace(" ", "").split(",")
    snackbox += snacks
    # return snackbox

# 간식박스 표시
def display_snackbox(snackbox) :
    print("간식박스:", snackbox) 
    # return snackbox

# 간식박스 제거함수
def remove_snackbox(snackbox) :
    snacks = input("삭제하고 싶은 간식을 추가하세요. (쉼표로 분리) :").replace(" ", "").split(",")
    print("삭제 간식:", snacks)
    for snack in snacks :
        snackbox.remove(snack)
    # snackbox = [snack for snack in snackbox if snack not in snacks]
    # return snackbox

# 간식박스 주문 확정
def confirm_snackbox(snackbox) :
    # 간식박스 표시
    display_snackbox(snackbox)
    confirm = input("간식박스 주문을 확정하시겠습니까? (Y/N):")
    if (len(confirm) > 0 and confirm.upper()[0] == 'Y') :
        print("주문이 완료되었습니다!\n감사합니다")
        return True
    else :
        print("다시 확인후 주문하시기 바랍니다!")
        return False

# 메뉴 출력
def display_menu() :
    print("1. 간식박스 추가")
    print("2. 간식박스 확인")
    print("3. 간식 제거")
    print("4. 간식박스 확정")
    print("0. 나가기")
    
snackbox = make_snackbox()
print("간식박스 생성완료", snackbox)

while True :
    display_menu()
    menu_no = input("메뉴를 선택하세요")

    if menu_no == '1' :  # "1. 간식 추가"
        add_snackbox(snackbox)
    elif menu_no == '2' : # "2. 간식 확인"
        display_snackbox(snackbox) 
    elif  menu_no == '3' : # "3. 간식 삭제"
        remove_snackbox(snackbox)
    elif  menu_no == '4' : # "3. 간식 주문 확정"       
        if ( confirm_snackbox(snackbox) ) : break
    else :
        break
        
```

```python
snackbox = ['사과', '딸기', '떡볶이', '포도', '홈런볼', '두바이초콜릿','오뎅']

print(id(snackbox))

def remove_snackbox(snackbox) :
    print(id(snackbox))
    snacks = ['오뎅', '포도']
    snackbox = [snack for snack in snackbox if snack not in snacks]

    print("ID After comprehension :", id(snackbox))
    print(snackbox)

    return snackbox

#  snackbox = remove_snackbox(snackbox) 
remove_snackbox(snackbox) 
print("ID", id(snackbox))

print("-"*80)
print(snackbox)
```
