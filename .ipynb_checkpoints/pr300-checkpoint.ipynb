```python
# 35,36 % formatting
# 이름: 김민수 나이: 10
# 이름: 이철희 나이: 13

name1 = "김민수" 
age1 = 10
name2 = "이철희"
age2 = 13

format_str = "이름 : {0:%s}  나이 : {1:%d}"

print(format_str.format(name1, age1))
print(format_str.format(name2, age2))

print("이름: %s 나이: %d" % (name1, age1))
print("이름: %s 나이: %d" % (name2, age2))

format_str2 = "이름 : %s  나이 : %d"
print(format_str2 % (name1, age1))
print(format_str2 % (name2, age2))
```
