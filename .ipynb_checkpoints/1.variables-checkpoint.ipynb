```python
# 성장률 = (이번달 매출 - 지난달 매출) / 지난달 매출 * 100

sales_prev_month = 1240
sales_cur_month = 1370

avg_growth_rate = (sales_cur_month - sales_prev_month ) / sales_prev_month * 100

print("이번달 성장률은", f"{avg_growth_rate:.2f}"+"% 입니다.")
```

```python
print(f"이번달 성장률은 {avg_growth_rate:.2f}% 입니다.")
```

```python
def local_exam() :
    local_var = 100  # 로컬 변수
    print(f"함수 내부 : {local_var}")

local_var = 200  # 전역 변수, Global Variable
local_exam() # 함수 호출
print(f"외부 변수 : {local_var}")
```

```python
counter = 0

def int_counter() :
    global counter   # global 변수 선언
    # counter = counter + 1
    counter += 1
    print(counter)

int_counter()
int_counter()
int_counter()
```
