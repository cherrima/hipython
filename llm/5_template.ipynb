# ChatPromptTemplate 클래스

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

llm = ChatOpenAI(model="gpt-4o-mini")

chat_template = ChatPromptTemplate.from_messages([
    ("system", "당신은 베테랑 여행 가이드로, 고객 최적의 맞춤 여행 일정 수립을 도와줍니다."),
    ("human", "{when}에 {place}로 {travel}을 계획하고 있습니다."),
    ("ai",    "제가 바로 {place} 전문 여행 가이드입니다. 어떤 도움이 필요하신가요?"),
    ("human", "{question}")
])

prompt = chat_template.format_messages(
    when="여름",
    place="동남아",
    travel="가족여행",
    question="해당 지역의 여행지 5군데를 추천해 주세요"
)

# print(prompt)         # 메시지 리스트 확인

# invoke 호출 전 메시지 구성 확인
for i, msg in enumerate(prompt):
    print(f"[{i}] {type(msg).__name__}: {msg.content}")
    
response = llm.invoke(prompt)

print(f"\n[Final Response] {type(response).__name__}: {response.content}")
```

## LCEL 체인으로 리팩터링

```python
chain = chat_template | llm

response = chain.invoke({
    "when": "여름",
    "place": "동남아",
    "travel": "가족여행",
    "question": "해당 지역의 여행지 5군데를 추천해 주세요"
})
print(response.content)
```

## 프롬프트 컴포지션 (Prompt Composition)
- 여러 개의 프롬프트 조각을 조립해서 하나의 완성된 프롬프트를 만드는 기법입니다.

* 좋은 예 — 조각으로 분리 후 조립


| 페르소나 프롬프트 | + | 언어 규칙 | + | 형식 규칙 | + | 질문 |
| :--- | :--: | :--- | :--: | :--- | :--: | :--- |
| 재사용 가능 | | 재사용 가능 | | 재사용 가능 | | 매번 변경 |
  

```python
# 조각 정의
prompt_1 = PromptTemplate.from_template("당신은 {place} 베테랑 여행 가이드입니다. ")
prompt_2 = PromptTemplate.from_template("고객 최적의 {place} {travel} 일정 수립을 도와줍니다.")

# 입력값
inputs = {
    "place": "동남아",
    "travel": "가족여행"
}
# --- 방법 1: + 연산자 조합 ---
final_prompt = prompt_1 + prompt_2
result = final_prompt.format(**inputs)
print("[ + 연산자 결과 ]")
print(result)
# 당신은 동남아 베테랑 여행 가이드입니다. 고객 최적의 동남아 가족여행 일정 수립을 도와줍니다.
```

```python
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

# ================================================================
# 1. 금융 도메인 프롬프트 구성
# ================================================================

chat_template = ChatPromptTemplate.from_messages([
    ("system",  "당신은 레이달리오로, 고객에게 {포트폴리오} 전략을 쉽고 명확하게 설명해 드립니다."),
    ("human",   "저는 {level} 투자자로, {포트폴리오}에 관심이 많습니다."),
    ("ai",      "안녕하세요! {포트폴리오} 전문가 레이달리오입니다. {level} 눈높이에 맞춰 설명해 드리겠습니다. 무엇이 궁금하신가요?"),
    ("human",   "{question}")
])

# ================================================================
# 2. LCEL 파이프라인 구성
# ================================================================

llm = ChatOpenAI(model="gpt-4o-mini", temperature=0.3)
output_parser = StrOutputParser()

chain = chat_template | llm | output_parser

# ================================================================
# 3. invoke
# ================================================================

# 1. 전달할 변수 정의
input_data = {
    "포트폴리오": "여러 금융상품(채권, 주식, 원자재 선물 등)",
    "level": "중급자",
    "question": "현금 1억으로 연평균 6% 수익을 낼 수 있는 포트폴리오를 추천해주세요."
}

# 2. 실제로 LLM에 전달되기 직전의 Prompt 메시지 객체들을 출력
prompt_messages = chat_template.format_messages(**input_data)

print("----- [Prompt Messages] -----")
for i, msg in enumerate(prompt_messages):
    print(f"[{i}] {type(msg).__name__}: {msg.content}")
print("-"*120)

# 3. 체인 실행
response = chain.invoke(input_data)
print("----- [Final Response] -----")
print(response)

# response = chain.invoke({
#     "포트폴리오"  : "여러 금융상품(채권, 주식, 원자재 선물 등)",
#     "level"   : "중급자",
#     "question": "현금 1억으로 연평균 6% 수익을 낼 수 있는 포트폴리오를 추천해주세요."
# })

# # # invoke 호출 전 메시지 구성 확인
# # for i, msg in enumerate(prompt):
# #     print(f"[{i}] {type(msg).__name__}: {msg.content}")

# print(response)
```

```python
import yfinance as yf

import textwrap

from langchain_core.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate, AIMessagePromptTemplate

from langchain_core.output_parsers import StrOutputParser

from langchain_openai import ChatOpenAI

# 1. 실시간 주가 데이터 가져오는 함수

def get_stock_info(ticker_symbol):

    try:

        stock = yf.Ticker(ticker_symbol)

        data = stock.history(period="1d")

        current_price = data['Close'].iloc[-1]

        prev_close = stock.info.get('previousClose', current_price)

        change_pct = ((current_price - prev_close) / prev_close) * 100

        

        return f"현재가: {current_price:,.2f}, 전일대비: {change_pct:+.2f}%"

    except:

        return "데이터를 가져올 수 없는 잡주(자코)임"

# 2. 프롬프트 템플릿 정의 (System-User-AI-User)

chat_prompt = ChatPromptTemplate.from_messages([

    SystemMessagePromptTemplate.from_template(

        "당신은 금융 지식이 해박하지만 성격이 뒤틀린 '소악마 메스가키' 자산관리사입니다. "

        "사용자를 '허접(자코)', '오빠'라고 부르며 비웃지만, 제공된 '실시간 주가 데이터'를 근거로 "

        "냉철하게 뼈를 때려야 합니다. 문장은 읽기 편하게 적절히 줄바꿈을 섞어주세요. ♡"

    ),

    # Few-shot: 말투와 논리 학습

    HumanMessagePromptTemplate.from_template("나 삼성전자(005930.KS) 풀매수했어! 나 이제 부자 되는 거야?"),

    AIMessagePromptTemplate.from_template(

        "우와아~ 진짜 대책 없는 허접 오빠네? ♡ \n"

        "방금 확인해보니까 지금 이 종목, 흐름도 안 좋은데 거기다 전 재산을 태워? \n"

        "분산 투자라는 기초 상식도 없는 거야? 진짜 '자코' 그 자체네! \n"

        "당장 포트폴리오 재점검 안 하면 조만간 한강 물 온도 체크하게 될걸? 허~접! ♡"

    ),

    # 실제 입력

    HumanMessagePromptTemplate.from_template(

        "상태: {investor_status}\n종목명: {stock_name}\n실시간 데이터: {stock_data}\n\n"

        "위 정보를 바탕으로 이 오빠한테 한마디 해줘!"

    )

])

# 3. 체인 구성

model = ChatOpenAI(model="gpt-4o")

chain = chat_prompt | model | StrOutputParser()

# 4. 가독성을 위한 출력 함수 (셀 안에서 안 벗어나게 함)

def print_wrapped_result(text, width=60):

    print("="*width)

    print(f"       [ 😈 귀여운 여동생의 독설 조언 ]")

    print("="*width)

    # 한글 특성을 고려해 textwrap 적용

    wrapped_text = textwrap.fill(text, width=width, break_long_words=False)

    print(wrapped_text)

    print("="*width)

# 5. 실행

ticker = "WMT"  # 테슬라 예시

stock_data = get_stock_info(ticker)

inputs = {

    "investor_status": "미국 주식에 퇴직금을 올인한 상태",

    "stock_name": "월마트(WMT)",

    "stock_data": stock_data

}

result = chain.invoke(inputs)

print_wrapped_result(result)
```
